# Access Modifiers >> Public, Protected, Private

```python
# Public      age
# Protected   _age
# Private     __age
```

```python
# Public
class Student():
    def __init__(self):
        self.name="Amit"
        self.marks=85
s1=Student()
print(s1.name)
```

```python
# Protected
class Employee():
    def __init__(self):
        self.name="Priya"
        self._salary=20000

class Developer(Employee):
    def showDetails(self):
        print("Employee Name :",self.name)
        print("Employee Salary :",self._salary)

d1=Developer()
d1.showDetails()
print(d1._salary)
```

```python
# Private
class BankAccount():
    def __init__(self):
        self.__balance=20000

    def showDetails(self):
        print(self.__balance)

account=BankAccount()
account.showDetails()
print(account.__balance)
```

```python
class BankAccount():
    def addDetails(self,name,balance,pin):
        self.__name=name       # set
        self.__balance=balance # set
        self.__pin=pin           # set

    def showAccountDetails(self):
        print("User Name :",self.__name)        # Get
        print("User Balance :",self.__balance)  # Get
        print("User Pin :",self.__pin)          # Get

bank=BankAccount()
bank.addDetails("Rahul",2000,"ABCD123")
bank.showAccountDetails()
```

```python
class BankAccount():
    def setName(self,name):
        self.__name=name       # set
    def setBalance(self,balance):
        self.__balance=balance # set
    def setPin(self,pin):
        self.__pin=pin           # set

    def showAccountDetails(self):
        print("User Name :",self.__name)        # Get
        print("User Balance :",self.__balance)  # Get
        print("User Pin :",self.__pin)          # Get

bank=BankAccount()
bank.setName("Rahul")
bank.setBalance(2000)
bank.setPin("ABCD123")
bank.showAccountDetails()
```

```python
class BankAccount():
    def __init__(self,name,balance,pin):
        self.__name=name        # set
        self.__balance=balance  # set
        self.__pin=pin           # set
    def setName(self,name):
        self.__name=name       # set
    def setBalance(self,balance):
        self.__balance=balance # set
    def setPin(self,pin):
        self.__pin=pin           # set

    def showAccountDetails(self):
        print("User Name :",self.__name)        # Get
        print("User Balance :",self.__balance)  # Get
        print("User Pin :",self.__pin)          # Get

bank=BankAccount("Rahul",2000,"ABCD123")
bank.showAccountDetails()

print("="*25)
bank.setName("Rahul Singh")
bank.showAccountDetails()

print("="*25)
bank.setPin("1234567")
bank.showAccountDetails()
```

```python
# Complete Encapsulation Example
class BankAccount():
    def __init__(self,name,balance,pin):
        self.__name=name        # set
        self.__balance=balance  # set
        self.__pin=pin           # set
    def setName(self,name):
        self.__name=name       # setter
    def setBalance(self,balance):
        self.__balance=balance # setter
    def setPin(self,pin):
        self.__pin=pin           # setter

    def getName(self):
        return self.__name        # Getter
    def getBalance(self):
        return self.__balance     # Getter
    def getPin(self):
        return self.__pin         # Getter

obj1=BankAccount("Rahul",2000,"ABCD123")
print(obj1.getName())
print(obj1.getBalance())
print(obj1.getPin())
print("="*25)
```
