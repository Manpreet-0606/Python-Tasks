```python
def show():
    x=20
    print("Local Variable: ", x)
show()
```

```python
x=30  # Global Variable
def show():
    x=20  # Local Variable
    print("local Variable: ",x)
show()
print("Global Variable: ",x)
```

```python
x=30  # Global Variable
def show():
    global x
    print("local Variable: ",x)
show()
print("Global Variable: ",x)
```

```python
def display():
    global x
    print("Global Variable: ",x)
display()
```

```python
balance=3000
def show():
    global balance
    deposit=1000
    print("Deposit Balance: ",deposit)
    balance=balance+deposit
show()
print("Current Balance: ",balance)
```

```python
balance=3000
def show(): # Function without parameter
    global balance
    deposit=1000
    print("Deposit Balance: ",deposit)
    
    balance=balance+deposit  # 4000

print("Initial Balance: ",balance)
    
show()
print("Current Balance: ",balance)
```

```python
balance=3000
def show():
    deposit=1000
    print("Deposit Balance: ",deposit)

    print("Initial Balance was: ",globals() ["balance"]) # One time use
show()
```

```python
# Function with Parameter/arguments
def show(x,y):
    add=x+y
    print(add)

show(10,30)
show(13,2)
```

```python
# Function with Parameter/arguments but no return value
def add(x,y):
    z=x+y
    print(z)

add(10,20)
add(40,20)
add(10,3)
```

```python
# Function with Parameter/arguments with return value
def add(x,y):
    z=x+y
    return z
add(20,30)
add(20,10)
```

```python
def calculateBill(price,discount=0):
    discount_amount=price*discount/100
    bill=price-discount_amount
    print("=======================")
    print("Original Price: ",price)
    print("Discount: ",discount, "%")
    print("Discount: ",discount_amount)
    print("Final Bill: ",bill)
    print("=======================")

calculateBill(20000)  # Cash 00% Discount
calculateBill(10000,10)  # UPI 10% Discount
calculateBill(30000,20)  # Card 20% Discount
```

```python
# args / vargs
def add(*args): # Data store in Tuple Form
    total=0
    for ex in args:
        total=total+ex
    print("Sum: ",total)
add(10,20) # args=(10,20)
add(40,50,60) #args=(40,50,60)
add(20,30,40,50) #parameter=4
add(20,30,40,50,60) #parameter=5
```

```python
def add(*args):
    print("args[0]",args[0])
    print("args[1]",args[1])
    print("args[2]",args[2])
    print("args[3]",args[3])
    print("args[4]",args[4])
    print(args) #tuple(20, 30, 40, 50, 60)
                #list[20, 30, 40, 50, 60]
add(20,30,40,50,60)
```

```python
def add(*args):
    bill=0
    for price in args:
        bill=bill+price
    print("Bill: ",bill)
add(2999,1999,2599)
```

```python
# *kwargs 
# keyword Arguments >> Dictionary

def Employeedetails(**kwargs):
    print(kwargs)

Employeedetails(Name="Mohit",Role="Developer",Department="IT")
Employeedetails(Name="Mohit",Department="IT")
Employeedetails(Name="Mohit")

# kwargs={Name:"Mohit",Role:"Developer",Department:"IT"}
# kwargs={Name:"Mohit",Department:"IT"}
# kwargs={Name:"Mohit"}
```

```python
def Employeedetails(**kwargs):
    print(kwargs['Name'])
    print(kwargs['Role'])
    print(kwargs['Department'])

Employeedetails(Name="Mohit",Role="Developer",Department="IT")
```

```python
def Employeedetails(**kwargs):
    for key, value in kwargs.items():
        print(key,":",value)
    print("===============")

Employeedetails(Name="Mohit",Role="Developer",Department="IT")
Employeedetails(Name="Mohit",Department="IT")
Employeedetails(Name="Mohit")
```
