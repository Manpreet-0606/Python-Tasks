```python
def even_number(start, end):
    for num in range(start, end+1):
        if num%2==0:
            return num  # print only one no.

even_number(1,10)
```

```python
def even_number(start, end):
    for num in range(start, end+1):
        if num%2==0:
            yield num

x=even_number(1,10)  #2, 4, 6, 8, 1o
print(next(x,"Error : Iteration Over"))
print(next(x,"Error : Iteration Over"))
print(next(x,"Error : Iteration Over"))
print(next(x,"Error : Iteration Over"))
print(next(x,"Error : Iteration Over"))
print(next(x,"Error : Iteration Over"))
```

```python
def even_number(start, end):
    for num in range(start, end+1):
        if num%2==0:
            yield num  # print untill loop end.

for i in even_number(1,10):
    print(i)
```

```python
# By Generator
def get_numbers():
    yield [10, 20, 30]

for g in get_numbers():
    print(g)
```

```python
def get_numbers():
    yield (10, 20, 30)

for g in get_numbers():
    print(g)
```

```python
def get_numbers():
    yield 10
    yield 20
    yield 30

for g in get_numbers():
    print(g)
    print(type(g))
```

```python
def get_numbers():  # waiting
    yield 10
    yield 20 
    yield 30 # wait

ex=get_numbers()
print(next(ex)) #10
print(next(ex)) #20
print(type(ex))
```

```python
# Student Records
students = [
    {"Name": "Aman", "Marks": [80, 75, 90]},
    {"Name": "Riya", "Marks": [35, 40, 30]},
    {"Name": "Rahul", "Marks": [90, 85, 95]},
    {"Name": "Neha", "Marks": [60, 70, 65]}
]
def gen_result(student_record):
    for students in student_record:
        name=students['Name']
        marks=students['Marks']

        avg=sum(marks)/len(marks)
        if avg>=40:
            status="Pass"
        else:
            status="Fail"

        yield {
            "Name":name,
            "Average":avg,
            "Status":status
        }

for ex in gen_result(students):
    print(ex)
```

```python
students = [
    {"Name": "Aman", "Marks": [80, 75, 90]},
    {"Name": "Riya", "Marks": [35, 40, 30]},
    {"Name": "Rahul", "Marks": [90, 85, 95]},
    {"Name": "Neha", "Marks": [60, 70, 65]}
]
def gen_result(student_record):
    for students in student_record:
        name=students['Name']
        marks=students['Marks']

        avg=sum(marks)/len(marks)
        if avg>=40:
            status="Pass"
        else:
            status="Fail"

        yield {
            "Name":name,
            "Average":avg,
            "Status":status
        }

for result in gen_result(students):
    print(result['Name'])
    print(result['Average'])
    print(result['Status'])
    print('--------------------')
```
