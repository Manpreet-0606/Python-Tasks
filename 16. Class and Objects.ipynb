```python
# Constructors : __init__()
class Student:
    def __init__(self,name,age,course):
        self.n=name
        self.a=age
        self.c=course
    def showdetails(self):
        print(self.n)
        print(self.a)
        print(self.c)
        print("===========")

obj1=Student("Amit",21,"Python")
obj1.showdetails()

obj2=Student("Jhon",21,"Python")
obj2.showdetails()

obj3=Student("Priya",21,"Java")
obj3.showdetails()
```

```python
class Employee: # E_ID, E_Name, E_Salary, E_Company
    def addEmpDetails(self, E_ID, E_Name, E_Salary, E_Company):
        self.id=E_ID  # (E_ID is a local variable)
        self.name=E_Name
        self.salary=E_Salary
        self.company=E_Company
        # (self.company is a instance variable)
        
    def showEmpDetails(self):
        print("Employee ID: ",self.id)
        print("Employee Name: ",self.name)
        print("Employee Salary: ",self.salary)
        print("Employee Company: ",self.company)
        print("="*25)

e1=Employee()
e1.addEmpDetails(111,"Raj",12000.00,"HCL")
e1.showEmpDetails()

e2=Employee()
e2.addEmpDetails(222,"Ansh Chaudhary",22000.00,"HCL")
e2.showEmpDetails()
```
