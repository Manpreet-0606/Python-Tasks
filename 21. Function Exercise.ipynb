```python
# EXERCISE 3
def empProfile(**details):
    for key, value in details.items():
        print(key,":",value)
    print("===================")

empProfile(Name="Soniya",Department="HR",Salary=40000,City="Delhi",Experience="2 YEAR")
```

```python
# EXERCISE 4
def studentResult(name,*marks,**details):
    print("Student Name: ",name)
    print("Subject Marks: ",marks)
    
    if len(marks)>0:
        total=sum(marks)
        avg=total/len(marks)
        print("Total Marks: ",total)
        print("Avg Marks: ",avg)

    print("=================Additional Information================")
    for key, value in details.items():
        print(key,"%",value)
    print("=======================")

studentResult("Amit",Roll_no=101,Sem="1st")
studentResult("Rahul",50,80,49,66,97,Roll_no=102,Sem="1st")
```

```python
# EXERCISE 5
def addItems(*prices):
    print("Prices: ",prices)
    print("Total Price: ",sum(prices))
    
def CustomerDetails(Name,City,Member):
    print("NAME: ",Name)
    print("CITY: ",City)
    print("MEMBER: ",Member)
    
prices=[20,65,77,85,34]
addItems(*prices)

customer={
    "Name":"Aman",
    "City":"Delhi",
    "Member":"Gold"
}
CustomerDetails(**customer)
```

```python
# EXERCISE 6
def calculate_avg(*marks):
    if not marks:
        return "Error : No Marks Provided."
    for n in marks:
        if not isinstance(n,(int,float)) or isinstance(n,bool):
            return "ERROR : Marks Must be Numeric"
        if n<0 or n>100:
            return "ERROR : Marks must be between 0 and 100"
    return sum(marks)/len(marks)
    
print(calculate_avg(55,66,58,97,75))
print(calculate_avg())
print(calculate_avg(55,66,58,97,75,True))
print(calculate_avg(55,66,58,97,75,"HI"))
```
