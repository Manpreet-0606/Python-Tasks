# Abstract Method

```python
from abc import ABC,abstractmethod
class Example(ABC):
    @abstractmethod
    def show(self):
        pass
```

```python
from abc import ABC,abstractmethod
class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):
    def sound(self): # Method Overriding
        print("Dog Barks")

class Cat(Animal):
    def sound(self): # Method Overriding
        print("Cat Meows")

d1=Dog()
d1.sound()


c1=Cat()
c1.sound()
```
